```python
import matplotlib.pyplot as plt
import networkx as nx
import pandas as pd
import numpy as np

# =============================================
# DATA NODE (RUANGAN)
# =============================================
room_data = {
    1: {"id": "K-0", "nama": "Kantor", "inisial": "KR", "area": "Kantor"},
    2: {"id": "LP", "nama": "Lantai Pabrik", "inisial": "LPK", "area": "Produksi"},
    3: {"id": "LPRO-1", "nama": "Menara Air", "inisial": "MNR", "area": "Produksi"},
    4: {"id": "LPRO-2", "nama": "Gardu Listrik", "inisial": "GRD", "area": "Produksi"},
    5: {"id": "LPRO-3", "nama": "Kompresor", "inisial": "KOMP", "area": "Produksi"},
    6: {"id": "LPRO-4", "nama": "Generator", "inisial": "GEN", "area": "Produksi"},
    7: {"id": "LPRO-5", "nama": "Bengkel", "inisial": "BKL", "area": "Produksi"},
    8: {"id": "LPRO-6", "nama": "Area Limbah", "inisial": "LMBH", "area": "Produksi"},
    9: {"id": "LPAB-1", "nama": "Garasi", "inisial": "GRS", "area": "Parkir"},
    10: {"id": "LPAB-2", "nama": "Rumah Jaga", "inisial": "RJG", "area": "Parkir"},
    11: {"id": "LPAB-3", "nama": "Parkir Mobil", "inisial": "PKR", "area": "Parkir"},
    12: {"id": "LPER-1", "nama": "Mushola", "inisial": "MSL", "area": "Fasilitas"},
    13: {"id": "LPER-2", "nama": "Kantin & Dapur", "inisial": "PTR", "area": "Fasilitas"},
    14: {"id": "LPER-3", "nama": "Toilet Pria", "inisial": "WLK", "area": "Fasilitas"},
    15: {"id": "LPER-4", "nama": "Toilet Wanita", "inisial": "WPR", "area": "Fasilitas"},
    16: {"id": "LPER-5", "nama": "Wastafel", "inisial": "WAS", "area": "Fasilitas"},
    17: {"id": "LPER-6", "nama": "UKP", "inisial": "UKP", "area": "Fasilitas"},
    18: {"id": "LPER-7", "nama": "Loker & Ganti", "inisial": "LKR", "area": "Fasilitas"}
}

# =============================================
# DATA EDGE DARI ARC (SUDAH LENGKAP)
# =============================================

# 🔴 PRIORITAS TERTINGGI (A = 1.0)
edges_A = [
    (1,11), (2,3), (2,4), (2,5), (2,6), (2,7),
    (4,6), (10,11), (14,16), (15,16), (16,17)
]

# 🟠 SANGAT PENTING (E = 0.8)
edges_E = [
    (1,12), (1,13), (1,17), (2,8), (2,14), (2,15), (2,16), (2,18),
    (3,14), (3,15), (3,16), (7,8), (7,16), (12,1), (13,1), (14,2),
    (14,3), (15,2), (15,3), (16,2), (16,3), (16,7), (17,1), (18,2)
]

# 🟡 PENTING (I = 0.5)
edges_I = [
    (1,10), (1,18), (2,9), (3,6), (3,7), (3,8), (4,5), (4,7),
    (5,6), (5,7), (6,7), (9,11), (12,14), (12,15), (12,16),
    (13,14), (13,15), (13,16), (14,15), (14,17), (14,18),
    (15,17), (15,18), (16,18), (17,14), (17,15), (18,14),
    (18,15), (18,16)
]

# 🔵 CUKUP (O = 0.3)
edges_O = [
    (1,4), (2,11), (3,4), (5,8), (6,8), (7,14), (7,15), (7,17),
    (7,18), (8,9), (8,10), (10,14), (10,15), (10,16), (10,17),
    (11,12), (11,13), (12,13), (13,17), (17,7), (17,10), (18,7), (18,17)
]

# ⚫ TIDAK DIINGINKAN (X = -0.02)
edges_X = [
    (1,2), (1,5), (1,6), (1,8), (2,13), (2,17), (5,12), (5,17),
    (6,12), (6,17), (8,11), (8,12), (8,13), (8,14), (8,15), (8,16),
    (8,17), (8,18), (12,5), (12,6), (12,8), (13,2), (13,8), (14,8),
    (15,8), (16,8), (17,2), (17,5), (17,6), (17,8), (18,8)
]

# =============================================
# MEMBANGUN GRAPH
# =============================================
G = nx.Graph()

for u, v in edges_A:
    G.add_edge(u, v, weight=1.0, relation='A')
for u, v in edges_E:
    G.add_edge(u, v, weight=0.8, relation='E')
for u, v in edges_I:
    G.add_edge(u, v, weight=0.5, relation='I')
for u, v in edges_O:
    G.add_edge(u, v, weight=0.3, relation='O')
for u, v in edges_X:
    G.add_edge(u, v, weight=-0.02, relation='X')

# =============================================
# HITUNG WEIGHTED RANK (BUKAN DEGREE!)
# =============================================
weighted_rank = {}
for node in G.nodes():
    total_weight = 0
    for neighbor in G.neighbors(node):
        total_weight += G[node][neighbor]['weight']
    weighted_rank[node] = total_weight

# Sort berdasarkan weighted rank (tertinggi = paling penting)
sorted_by_weight = sorted(weighted_rank.items(), key=lambda x: x[1], reverse=True)

# =============================================
# WARNA NODE BERDASARKAN RANK WEIGHT
# =============================================
# Semakin tinggi weight, semakin merah gelap
# Semakin rendah (negatif), semakin biru
colors_weight = []
for node in G.nodes():
    w = weighted_rank[node]
    if w >= 5.0:
        colors_weight.append('#8B0000')  # Dark Red
    elif w >= 3.0:
        colors_weight.append('#DC143C')  # Crimson
    elif w >= 1.0:
        colors_weight.append('#FF6347')  # Tomato
    elif w >= 0:
        colors_weight.append('#FFA07A')  # Light Salmon
    else:
        colors_weight.append('#4682B4')  # Steel Blue (negatif)

# =============================================
# LAYOUT
# =============================================
pos = nx.spring_layout(G, seed=42, k=2.5, iterations=100)

# =============================================
# VISUALISASI
# =============================================
fig, ax = plt.subplots(figsize=(24, 18))

# Draw nodes dengan warna berdasarkan weight rank
nx.draw_networkx_nodes(G, pos, node_size=1600, node_color=colors_weight,
                       edgecolors='black', linewidths=2.5, alpha=0.95)

# Draw edges
nx.draw_networkx_edges(G, pos, edgelist=edges_A, width=5.0, alpha=0.95,
                       edge_color='#FF0000', style='solid')
nx.draw_networkx_edges(G, pos, edgelist=edges_E, width=3.5, alpha=0.85,
                       edge_color='#FF8C00', style='solid')
nx.draw_networkx_edges(G, pos, edgelist=edges_I, width=2.5, alpha=0.75,
                       edge_color='#32CD32', style='solid')
nx.draw_networkx_edges(G, pos, edgelist=edges_O, width=1.5, alpha=0.65,
                       edge_color='#4169E1', style='dashed')
nx.draw_networkx_edges(G, pos, edgelist=edges_X, width=1.5, alpha=0.5,
                       edge_color='#2F4F4F', style='dotted')

# Labels
labels = {n: f"{room_data[n]['nama']}\n({room_data[n]['inisial']})\n{weighted_rank[n]:.1f}"
          for n in G.nodes()}
nx.draw_networkx_labels(G, pos, labels, font_size=7, font_weight='bold')

# =============================================
# LEGEND
# =============================================
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

legend_elements = [
    # Relationship types
    Line2D([0], [0], color='#FF0000', lw=5.0, label='A - Absolutely Necessary (1.0)'),
    Line2D([0], [0], color='#FF8C00', lw=3.5, label='E - Especially Important (0.8)'),
    Line2D([0], [0], color='#32CD32', lw=2.5, label='I - Important (0.5)'),
    Line2D([0], [0], color='#4169E1', lw=1.5, linestyle='dashed', label='O - Ordinary (0.3)'),
    Line2D([0], [0], color='#2F4F4F', lw=1.5, linestyle='dotted', label='X - Undesirable (-0.02)'),
    # Weight rank colors
    Patch(facecolor='#8B0000', edgecolor='black', label='Rank Weight: > 5.0 (Sangat Tinggi)'),
    Patch(facecolor='#DC143C', edgecolor='black', label='Rank Weight: 3.0 - 5.0 (Tinggi)'),
    Patch(facecolor='#FF6347', edgecolor='black', label='Rank Weight: 1.0 - 3.0 (Sedang)'),
    Patch(facecolor='#FFA07A', edgecolor='black', label='Rank Weight: 0 - 1.0 (Rendah)'),
    Patch(facecolor='#4682B4', edgecolor='black', label='Rank Weight: < 0 (Negatif)'),
]

ax.legend(handles=legend_elements, loc='upper left', bbox_to_anchor=(1.02, 1),
          fontsize=9, frameon=True, fancybox=True, shadow=True)

plt.title("ARC Network Graph - Layout Pabrik\nRanking Berdasarkan TOTAL WEIGHT (Prioritas Hubungan)",
          fontsize=16, fontweight='bold', pad=20)
plt.axis('off')
plt.tight_layout()
plt.show()

# =============================================
# RANKING BERDASARKAN WEIGHT (BUKAN DEGREE!)
# =============================================
print("="*80)
print("🏆 RANKING RUANGAN BERDASARKAN TOTAL WEIGHT (PRIORITAS HUBUNGAN)")
print("="*80)
print(f"{'Rank':<6} {'No':<4} {'Nama Ruangan':<25} {'Inisial':<8} {'Total Weight':<12} {'Interpretasi'}")
print("-"*80)

for rank, (node, weight) in enumerate(sorted_by_weight, 1):
    nama = room_data[node]['nama']
    inisial = room_data[node]['inisial']

    if weight >= 5.0:
        interpretasi = "🔴 PRIORITAS TERTINGGI - Pusat aktivitas"
    elif weight >= 3.0:
        interpretasi = "🟠 PRIORITAS TINGGI - Sangat terhubung"
    elif weight >= 1.0:
        interpretasi = "🟡 PRIORITAS SEDANG - Cukup terhubung"
    elif weight >= 0:
        interpretasi = "🔵 PRIORITAS RENDAH - Minim keterhubungan"
    else:
        interpretasi = "⚫ PRIORITAS NEGATIF - Harus dijauhkan"

    print(f"{rank:<6} {node:<4} {nama:<25} {inisial:<8} {weight:<12.2f} {interpretasi}")

print("\n" + "="*80)
print("📊 ANALISIS PRIORITAS LAYOUT:")
print("="*80)

# Cari node dengan weight tertinggi
top_node = sorted_by_weight[0][0]
top_weight = sorted_by_weight[0][1]
print(f"\n🎯 RUANGAN PALING KRITIS (Total Weight Tertinggi):")
print(f"   → {room_data[top_node]['nama']} ({room_data[top_node]['inisial']})")
print(f"   → Total Weight: {top_weight:.2f}")
print(f"   → Harus diposisikan di area SENTRAL karena punya banyak hubungan PRIORITAS TINGGI")

# Cari node dengan weight negatif
neg_nodes = [(n, w) for n, w in weighted_rank.items() if w < 0]
if neg_nodes:
    print(f"\n⚠️ RUANGAN DENGAN WEIGHT NEGATIF:")
    for node, weight in neg_nodes:
        print(f"   → {room_data[node]['nama']} ({room_data[node]['inisial']}): {weight:.2f}")
        print(f"      Harus dijauhkan dari ruangan lain (prioritas X)")

print("\n" + "="*80)
print("💡 REKOMENDASI LAYOUT:")
print("="*80)
print("1. Letakkan ruangan dengan TOTAL WEIGHT TERTINGGI di POSISI SENTRAL")
print("2. Prioritaskan edge A (merah) untuk didekatkan terlebih dahulu")
print("3. Edge X (hitam putus-putus) harus dijauhkan SEMAKSIMAL MUNGKIN")
print("4. Gunakan Total Weight sebagai acuan utama penentuan posisi")
print("="*80)
```

```python
import numpy as np
import random
import math
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import pandas as pd

# =============================================
# DATA RUANGAN (dengan allowance 50%)
# =============================================
rooms_spec = [
    (1, "Direktur", 43.5, 1.33),
    (2, "Keuangan", 43.5, 1.67),
    (3, "SDM", 43.5, 1.25),
    (4, "Produksi", 40.5, 1.67),
    (5, "QC", 36.0, 1.67),
    (6, "Pemasaran", 36.0, 1.67),
    (7, "R&D", 28.5, 2.50),
    (8, "Rapat", 72.0, 1.33),
    (9, "Arsip", 22.5, 1.67),
    (10, "Resepsionis", 24.0, 1.00),
    (11, "Musholla", 30.0, 1.25),
    (12, "Pantry", 24.0, 1.00),
    (13, "Keamanan", 13.5, 1.00),
    (14, "Kesehatan", 13.5, 1.00),
    (15, "Toilet W", 4.5, 0.75),
    (16, "Toilet P", 4.5, 0.75),
    (17, "Sink", 3.6, 0.833)
]

rooms_data = []
for id_, name, area, aspect in rooms_spec:
    l_dim = np.sqrt(area / aspect)
    p_dim = aspect * l_dim
    rooms_data.append({
        "id": id_ - 1,  # 0-indexed
        "name": name,
        "p": p_dim,
        "l": l_dim,
        "area": area
    })

n = len(rooms_data)
room_names = [r['name'] for r in rooms_data]

# =============================================
# FLOW MATRIX (W) - Real ARC data
# =============================================
W = np.zeros((n, n))
arc_edges = [
    (1,2,0.8), (1,3,0.8), (1,4,0.5), (1,5,0.5), (1,6,0.8), (1,7,0.5), (1,8,1.0), (1,9,0.3), (1,10,1.0), (1,13,0.3),
    (2,3,0.5), (2,4,0.5), (2,5,0.3), (2,6,0.5), (2,7,0.3), (2,8,0.5), (2,9,0.8), (2,10,0.3),
    (3,4,0.5), (3,5,0.3), (3,6,0.3), (3,7,0.3), (3,8,0.5), (3,9,0.8), (3,10,0.3), (3,13,0.5), (3,14,0.5), (3,15,0.5),
    (4,5,1.0), (4,6,0.5), (4,7,0.5), (4,8,0.5), (4,9,0.5), (4,14,0.3),
    (5,6,0.3), (5,7,0.5), (5,8,0.5), (5,9,0.5), (5,14,0.3), (5,15,0.3),
    (6,7,0.5), (6,8,0.8), (6,9,0.3), (6,10,0.8), (6,15,0.3),
    (7,8,0.5), (7,9,0.8), (7,10,0.3), (7,15,0.3),
    (8,9,0.5), (8,10,0.5), (8,12,0.8), (8,15,0.5), (8,16,0.5), (8,17,0.3),
    (9,12,-0.02), (9,13,0.3), (9,15,-0.02), (9,16,-0.02), (9,17,-0.02),
    (10,12,0.3), (10,13,1.0), (10,15,0.5), (10,16,0.5), (10,17,0.3),
    (11,15,0.3), (11,16,0.3), (11,17,1.0),
    (12,14,0.3), (12,15,0.5), (12,16,0.5), (12,17,1.0),
    (13,14,0.3),
    (14,15,0.5), (14,16,0.5), (14,17,1.0),
    (15,16,0.5), (15,17,0.5),
    (16,17,0.5)
]
for u, v, w in arc_edges:
    W[u-1, v-1] = w
    W[v-1, u-1] = w

print("✅ Flow Matrix W loaded")
print(f"   - Positive flows: {np.sum(W > 0)}")
print(f"   - Negative flows (X): {np.sum(W < 0)}")

# =============================================
# GRID DEFINITION
# =============================================
# Gedung: kita tentukan grid yang cukup besar
# Misal: 8x6 grid = 48 posisi (lebih dari cukup untuk 17 ruangan)

grid_rows = 8
grid_cols = 6
total_cells = grid_rows * grid_cols

# Ukuran tiap cell (dalam meter)
# Gedung total: 30m x 22m (sesuai perhitungan sebelumnya)
building_length = 30.0  # x
building_width = 22.0   # y

cell_width = building_length / grid_cols
cell_height = building_width / grid_rows

# Generate semua posisi grid
grid_positions = []
for r in range(grid_rows):
    for c in range(grid_cols):
        # Center of cell
        x = (c + 0.5) * cell_width
        y = (r + 0.5) * cell_height
        grid_positions.append((x, y, r, c))

# Kita hanya perlu n posisi (sesuai jumlah ruangan)
# Pilih posisi yang valid (nanti akan di-assign ke ruangan)
print(f"\n📐 Grid: {grid_rows}×{grid_cols} = {total_cells} cells")
print(f"   Cell size: {cell_width:.2f}m × {cell_height:.2f}m")

# =============================================
# DISTANCE MATRIX (Manhattan antar cell center)
# =============================================
def manhattan_distance(pos1, pos2):
    """Manhattan distance antara dua cell"""
    r1, c1 = pos1[2], pos1[3]
    r2, c2 = pos2[2], pos2[3]
    return (abs(r1 - r2) + abs(c1 - c2)) * ((cell_width + cell_height) / 2)  # approx meter

# Precompute distance matrix untuk semua cell
D_cells = np.zeros((total_cells, total_cells))
for i in range(total_cells):
    for j in range(total_cells):
        D_cells[i, j] = manhattan_distance(grid_positions[i], grid_positions[j])

# =============================================
# COST FUNCTION
# =============================================
def calculate_cost(assignment):
    """
    assignment: list of cell indices untuk setiap ruangan
    """
    total_cost = 0
    for i in range(n):
        for j in range(i+1, n):
            if W[i, j] != 0:
                dist = D_cells[assignment[i], assignment[j]]
                if W[i, j] > 0:
                    total_cost += W[i, j] * dist
                else:
                    # X relationship: reward kalau jauh (max 20m)
                    total_cost += abs(W[i, j]) * max(0, 20 - dist)
    return total_cost

# =============================================
# CONSTRAINT: Apakah assignment valid?
# =============================================
def is_valid_assignment(assignment):
    """Cek apakah tidak ada ruangan yang overlap dan muat di cell"""
    # Untuk simplifikasi, kita asumsikan tiap ruangan muat di 1 cell
    # (ini reasonable karena cell size ~3.75m × 2.75m, cukup untuk sebagian besar ruangan)

    # Cek apakah ada ruangan besar yang perlu multiple cells
    # Ruang Rapat (8) = 72m², perlu 2x2 cells?
    # Kita bisa handle secara terpisah, tapi untuk simplifikasi:

    # Cek apakah semua assignment unik
    return len(set(assignment)) == n

# =============================================
# INITIAL SOLUTION (Flow-based)
# =============================================
def initialize_flow_based():
    """Assign ruangan ke grid berdasarkan flow similarity"""
    from scipy.sparse.csgraph import laplacian
    from scipy.sparse.linalg import eigsh

    # Spectral ordering untuk dapatkan posisi relatif
    lap = laplacian(np.abs(W), normed=False)
    _, vecs = eigsh(lap, k=2, which='SM')
    coords = vecs[:, :2]

    # Normalize ke grid coordinates
    x_norm = (coords[:, 0] - coords[:, 0].min()) / (coords[:, 0].max() - coords[:, 0].min() + 1e-6)
    y_norm = (coords[:, 1] - coords[:, 1].min()) / (coords[:, 1].max() - coords[:, 1].min() + 1e-6)

    # Map ke grid indices
    assignment = []
    used_cells = set()

    for i in range(n):
        # Cari cell terdekat dengan koordinat normalized
        target_r = int(y_norm[i] * (grid_rows - 1))
        target_c = int(x_norm[i] * (grid_cols - 1))

        # Cari cell terdekat yang belum dipakai
        best_cell = None
        best_dist = float('inf')

        for cell_idx, (_, _, r, c) in enumerate(grid_positions):
            if cell_idx not in used_cells:
                dist = abs(r - target_r) + abs(c - target_c)
                if dist < best_dist:
                    best_dist = dist
                    best_cell = cell_idx

        assignment.append(best_cell)
        used_cells.add(best_cell)

    return assignment

# =============================================
# SIMULATED ANNEALING
# =============================================
def simulated_annealing(max_iter=5000):
    # Initial solution
    current = initialize_flow_based()
    current_cost = calculate_cost(current)

    best = current[:]
    best_cost = current_cost

    # SA parameters
    T_start = 1000.0
    T_end = 0.1
    alpha = (T_end / T_start) ** (1.0 / max_iter)
    T = T_start

    # Tracking
    costs_history = []

    for iteration in range(max_iter):
        # Generate neighbor: swap dua ruangan
        i, j = random.sample(range(n), 2)
        new = current[:]
        new[i], new[j] = new[j], new[i]

        new_cost = calculate_cost(new)
        delta = new_cost - current_cost

        # Accept?
        if delta < 0 or random.random() < math.exp(-delta / T):
            current = new
            current_cost = new_cost

            if current_cost < best_cost:
                best = current[:]
                best_cost = current_cost

        costs_history.append(current_cost)
        T *= alpha

        # Progress
        if iteration % 500 == 0:
            print(f"   Iter {iteration}: Best={best_cost:.2f}, Current={current_cost:.2f}, T={T:.3f}")

    return best, best_cost, costs_history

# =============================================
# RUN OPTIMIZATION
# =============================================
print("\n" + "="*60)
print("🔥 SIMULATED ANNEALING - FACILITY LAYOUT")
print("="*60)

best_assignment, best_cost, history = simulated_annealing(max_iter=3000)

print(f"\n✅ Optimization Complete!")
print(f"   Best Cost: {best_cost:.2f}")

# =============================================
# VISUALIZATION
# =============================================
fig, axes = plt.subplots(1, 2, figsize=(18, 8))

# Plot 1: Cost convergence
ax1 = axes[0]
ax1.plot(history)
ax1.set_xlabel('Iteration')
ax1.set_ylabel('Cost')
ax1.set_title(f'Simulated Annealing Convergence\nFinal Cost: {best_cost:.2f}')
ax1.grid(True, alpha=0.3)
ax1.set_yscale('log')

# Plot 2: Layout
ax2 = axes[1]
colors = plt.cm.tab20(np.linspace(0, 1, n))

# Draw grid cells
for cell_idx, (x, y, r, c) in enumerate(grid_positions):
    rect = patches.Rectangle(
        (x - cell_width/2, y - cell_height/2),
        cell_width, cell_height,
        linewidth=0.5, edgecolor='gray',
        facecolor='none', alpha=0.3
    )
    ax2.add_patch(rect)

# Draw rooms
room_positions = {}
for i, cell_idx in enumerate(best_assignment):
    x, y, r, c = grid_positions[cell_idx]
    room_positions[i] = (x, y)

    rect = patches.Rectangle(
        (x - cell_width/2.2, y - cell_height/2.2),
        cell_width/1.1, cell_height/1.1,
        linewidth=2, edgecolor='black',
        facecolor=colors[i], alpha=0.8
    )
    ax2.add_patch(rect)

    short_name = room_names[i].replace('Departemen ', '')[:10]
    ax2.text(x, y, f"{i+1}\n{short_name}",
             ha='center', va='center', fontsize=7, fontweight='bold')

# Draw flow lines (top 10)
edges_flow = []
for i in range(n):
    for j in range(i+1, n):
        if W[i, j] > 0.3:
            edges_flow.append((i, j, W[i, j]))

edges_flow.sort(key=lambda x: x[2], reverse=True)

for i, j, w in edges_flow[:10]:
    if w >= 1.0:
        color, width, alpha = 'red', 3.0, 0.6
    elif w >= 0.8:
        color, width, alpha = 'orange', 2.5, 0.5
    elif w >= 0.5:
        color, width, alpha = 'green', 2.0, 0.4
    else:
        color, width, alpha = 'blue', 1.5, 0.3

    x1, y1 = room_positions[i]
    x2, y2 = room_positions[j]
    ax2.plot([x1, x2], [y1, y2],
             color=color, linewidth=width, alpha=alpha, zorder=1)

ax2.set_xlim(0, building_length)
ax2.set_ylim(0, building_width)
ax2.set_aspect('equal')
ax2.set_title(f'Optimized Layout (Simulated Annealing)\nGrid: {grid_rows}×{grid_cols} | Cost: {best_cost:.2f}')
ax2.grid(True, alpha=0.3)

# Legend
import matplotlib.patches as mpatches
legend_elements = [
    mpatches.Patch(color='red', label='A (1.0)'),
    mpatches.Patch(color='orange', label='E (0.8)'),
    mpatches.Patch(color='green', label='I (0.5)'),
    mpatches.Patch(color='blue', label='O (0.3)'),
]
ax2.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# =============================================
# FINAL REPORT
# =============================================
print("\n" + "="*60)
print("📊 FINAL LAYOUT ASSIGNMENT")
print("="*60)
for i, cell_idx in enumerate(best_assignment):
    x, y, r, c = grid_positions[cell_idx]
    print(f"{i+1:2d}. {room_names[i]:<20} → Cell ({r},{c}) at ({x:.1f}, {y:.1f})")

# Check relationships
print("\n🔗 Top 5 Strongest Relationships:")
edges_flow.sort(key=lambda x: x[2], reverse=True)
for i, j, w in edges_flow[:5]:
    dist = D_cells[best_assignment[i], best_assignment[j]]
    print(f"   {room_names[i]} ↔ {room_names[j]}: weight={w:.1f}, distance={dist:.1f}m")
```

```python
import numpy as np
import random
import math
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# =============================================
# DATA RUANGAN (dengan allowance 50%)
# =============================================
rooms_spec = [
    (1, "Direktur", 43.5, 1.33),
    (2, "Keuangan", 43.5, 1.67),
    (3, "SDM", 43.5, 1.25),
    (4, "Produksi", 40.5, 1.67),
    (5, "QC", 36.0, 1.67),
    (6, "Pemasaran", 36.0, 1.67),
    (7, "R&D", 28.5, 2.50),
    (8, "Rapat", 72.0, 1.33),
    (9, "Arsip", 22.5, 1.67),
    (10, "Resepsionis", 24.0, 1.00),
    (11, "Musholla", 30.0, 1.25),
    (12, "Pantry", 24.0, 1.00),
    (13, "Keamanan", 13.5, 1.00),
    (14, "Kesehatan", 13.5, 1.00),
    (15, "Toilet W", 4.5, 0.75),
    (16, "Toilet P", 4.5, 0.75),
    (17, "Sink", 3.6, 0.833)
]

rooms = []
for id_, name, area, aspect in rooms_spec:
    l_dim = np.sqrt(area / aspect)  # lebar (y)
    p_dim = aspect * l_dim          # panjang (x)
    rooms.append({
        "id": id_ - 1,
        "name": name,
        "p": p_dim,
        "l": l_dim,
        "area": area,
        "rotated": False
    })

n = len(rooms)
room_names = [r['name'] for r in rooms]

# =============================================
# FLOW MATRIX (W) - Real ARC data (FIXED)
# =============================================
W = np.zeros((n, n))
arc_edges = [
    (1,2,0.8), (1,3,0.8), (1,4,0.5), (1,5,0.5), (1,6,0.8), (1,7,0.5), (1,8,1.0), (1,9,0.3), (1,10,1.0), (1,13,0.3),
    (2,3,0.5), (2,4,0.5), (2,5,0.3), (2,6,0.5), (2,7,0.3), (2,8,0.5), (2,9,0.8), (2,10,0.3),
    (3,4,0.5), (3,5,0.3), (3,6,0.3), (3,7,0.3), (3,8,0.5), (3,9,0.8), (3,10,0.3), (3,13,0.5), (3,14,0.5), (3,15,0.5),
    (4,5,1.0), (4,6,0.5), (4,7,0.5), (4,8,0.5), (4,9,0.5), (4,14,0.3),
    (5,6,0.3), (5,7,0.5), (5,8,0.5), (5,9,0.5), (5,14,0.3), (5,15,0.3),
    (6,7,0.5), (6,8,0.8), (6,9,0.3), (6,10,0.8), (6,15,0.3),
    (7,8,0.5), (7,9,0.8), (7,10,0.3), (7,15,0.3),
    (8,9,0.5), (8,10,0.5), (8,12,0.8), (8,15,0.5), (8,16,0.5), (8,17,0.3),
    (9,12,-0.02), (9,13,0.3), (9,15,-0.02), (9,16,-0.02), (9,17,-0.02),
    (10,12,0.3), (10,13,1.0), (10,15,0.5), (10,16,0.5), (10,17,0.3),
    (11,15,0.3), (11,16,0.3), (11,17,1.0),
    (12,14,0.3), (12,15,0.5), (12,16,0.5), (12,17,1.0),
    (13,14,0.3),
    (14,15,0.5), (14,16,0.5), (14,17,1.0),
    (15,16,0.5), (15,17,0.5),
    (16,17,0.5)
]
for u, v, w in arc_edges:
    W[u-1, v-1] = w
    W[v-1, u-1] = w

print(f"Data loaded: {n} rooms, total area = {sum(r['area'] for r in rooms):.1f} m²")

# =============================================
# INITIAL LAYOUT: Shelf Packing
# =============================================
def shelf_packing(rooms, building_length=None, building_width=None):
    """Pack rooms into shelves, return placements and building dimensions"""
    rooms_sorted = sorted(rooms, key=lambda r: r['l'], reverse=True)

    if building_length is None:
        total_area = sum(r['area'] for r in rooms)
        aspect = 1.5
        building_width = np.sqrt(total_area / aspect)
        building_length = building_width * aspect

    placements = []
    shelves = []  # (y_start, y_end, x_current)

    for room in rooms_sorted:
        placed = False

        for shelf_idx, (y_start, y_end, x_current) in enumerate(shelves):
            shelf_height = y_end - y_start

            # Try original orientation
            if room['l'] <= shelf_height and x_current + room['p'] <= building_length:
                placements.append({
                    'id': room['id'],
                    'name': room['name'],
                    'x': x_current,
                    'y': y_start,
                    'p': room['p'],
                    'l': room['l'],
                    'rotated': False
                })
                shelves[shelf_idx] = (y_start, y_end, x_current + room['p'] + 0.2)
                placed = True
                break

            # Try rotated
            if room['p'] <= shelf_height and x_current + room['l'] <= building_length:
                placements.append({
                    'id': room['id'],
                    'name': room['name'],
                    'x': x_current,
                    'y': y_start,
                    'p': room['l'],
                    'l': room['p'],
                    'rotated': True
                })
                shelves[shelf_idx] = (y_start, y_end, x_current + room['l'] + 0.2)
                placed = True
                break

        if not placed:
            if shelves:
                new_y = shelves[-1][1] + 0.2
            else:
                new_y = 0.2

            if new_y + room['l'] <= building_width:
                shelves.append([new_y, new_y + room['l'], 0.2])
                placements.append({
                    'id': room['id'],
                    'name': room['name'],
                    'x': 0.2,
                    'y': new_y,
                    'p': room['p'],
                    'l': room['l'],
                    'rotated': False
                })
                shelves[-1][2] = 0.2 + room['p'] + 0.2
            elif new_y + room['p'] <= building_width:
                shelves.append([new_y, new_y + room['p'], 0.2])
                placements.append({
                    'id': room['id'],
                    'name': room['name'],
                    'x': 0.2,
                    'y': new_y,
                    'p': room['l'],
                    'l': room['p'],
                    'rotated': True
                })
                shelves[-1][2] = 0.2 + room['l'] + 0.2
            else:
                building_width = new_y + max(room['l'], room['p']) + 0.5
                shelves.append([new_y, new_y + room['l'], 0.2])
                placements.append({
                    'id': room['id'],
                    'name': room['name'],
                    'x': 0.2,
                    'y': new_y,
                    'p': room['p'],
                    'l': room['l'],
                    'rotated': False
                })
                shelves[-1][2] = 0.2 + room['p'] + 0.2

    return placements, building_length, building_width

# =============================================
# COST FUNCTION
# =============================================
def calculate_cost(placements):
    pos = {}
    for p in placements:
        idx = p['id']
        pos[idx] = (p['x'] + p['p']/2, p['y'] + p['l']/2)

    total_cost = 0
    for i in range(n):
        for j in range(i+1, n):
            if W[i, j] != 0:
                x1, y1 = pos[i]
                x2, y2 = pos[j]
                dist = abs(x1 - x2) + abs(y1 - y2)
                if W[i, j] > 0:
                    total_cost += W[i, j] * dist
                else:
                    total_cost += abs(W[i, j]) * max(0, 20 - dist)
    return total_cost

# =============================================
# SWAP FUNCTION
# =============================================
def swap_rooms(placements, i, j):
    new_placements = []
    for p in placements:
        new_p = p.copy()
        if p['id'] == i:
            for q in placements:
                if q['id'] == j:
                    new_p['x'] = q['x']
                    new_p['y'] = q['y']
                    new_p['p'] = q['p']
                    new_p['l'] = q['l']
                    new_p['rotated'] = q['rotated']
                    break
        elif p['id'] == j:
            for q in placements:
                if q['id'] == i:
                    new_p['x'] = q['x']
                    new_p['y'] = q['y']
                    new_p['p'] = q['p']
                    new_p['l'] = q['l']
                    new_p['rotated'] = q['rotated']
                    break
        new_placements.append(new_p)
    return new_placements

# =============================================
# SIMULATED ANNEALING
# =============================================
def simulated_annealing(initial_placements, max_iter=2000):
    current = initial_placements
    current_cost = calculate_cost(current)

    best = current
    best_cost = current_cost

    T = 500.0
    alpha = 0.995
    costs_history = []

    for iteration in range(max_iter):
        i, j = random.sample(range(n), 2)
        new = swap_rooms(current, i, j)

        new_cost = calculate_cost(new)
        delta = new_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = new
            current_cost = new_cost

            if current_cost < best_cost:
                best = current
                best_cost = current_cost

        costs_history.append(current_cost)
        T *= alpha

        if iteration % 400 == 0:
            print(f"   Iter {iteration}: Best={best_cost:.2f}, Current={current_cost:.2f}")

    return best, best_cost, costs_history

# =============================================
# RUN
# =============================================
print("\n" + "="*60)
print("SHELF PACKING + SIMULATED ANNEALING")
print("="*60)

placements, bldg_l, bldg_w = shelf_packing(rooms)
initial_cost = calculate_cost(placements)

print(f"\nInitial layout (Shelf Packing):")
print(f"   Building: {bldg_l:.1f}m × {bldg_w:.1f}m")
print(f"   Area: {bldg_l * bldg_w:.1f} m²")
print(f"   Cost: {initial_cost:.2f}")

print("\n Simulated Annealing refinement...")
best_placements, best_cost, history = simulated_annealing(placements, max_iter=2000)

print(f"\n Optimized layout:")
print(f"   Cost: {best_cost:.2f} (improvement: {initial_cost - best_cost:.2f})")

# =============================================
# VISUALIZATION
# =============================================
fig, axes = plt.subplots(1, 2, figsize=(18, 8))

# Convergence plot
ax1 = axes[0]
ax1.plot(history)
ax1.axhline(y=initial_cost, color='r', linestyle='--', label=f'Initial: {initial_cost:.1f}')
ax1.set_xlabel('Iteration')
ax1.set_ylabel('Cost')
ax1.set_title('Simulated Annealing Convergence')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Layout plot
ax2 = axes[1]
colors = plt.cm.tab20(np.linspace(0, 1, n))

for p in best_placements:
    i = p['id']
    rect = patches.Rectangle(
        (p['x'], p['y']),
        p['p'], p['l'],
        linewidth=2, edgecolor='black',
        facecolor=colors[i], alpha=0.8
    )
    ax2.add_patch(rect)

    short_name = p['name'].replace('Departemen ', '')[:10]
    label = f"{i+1}\n{short_name}"
    if p['rotated']:
        label += "\n↻"
    ax2.text(p['x'] + p['p']/2, p['y'] + p['l']/2, label,
             ha='center', va='center', fontsize=7, fontweight='bold')

# Flow lines
pos_center = {p['id']: (p['x'] + p['p']/2, p['y'] + p['l']/2) for p in best_placements}

edges_flow = []
for i in range(n):
    for j in range(i+1, n):
        if W[i, j] > 0.3:
            edges_flow.append((i, j, W[i, j]))

edges_flow.sort(key=lambda x: x[2], reverse=True)

for i, j, w in edges_flow[:10]:
    if w >= 1.0:
        color, width, alpha = 'red', 3.0, 0.6
    elif w >= 0.8:
        color, width, alpha = 'orange', 2.5, 0.5
    elif w >= 0.5:
        color, width, alpha = 'green', 2.0, 0.4
    else:
        color, width, alpha = 'blue', 1.5, 0.3

    x1, y1 = pos_center[i]
    x2, y2 = pos_center[j]
    ax2.plot([x1, x2], [y1, y2],
             color=color, linewidth=width, alpha=alpha, zorder=1)

ax2.set_xlim(-0.5, bldg_l + 0.5)
ax2.set_ylim(-0.5, bldg_w + 0.5)
ax2.set_aspect('equal')
ax2.set_title(f'Optimized Layout\nBuilding: {bldg_l:.1f}×{bldg_w:.1f}m | Cost: {best_cost:.2f}')
ax2.grid(True, alpha=0.3)

import matplotlib.patches as mpatches
legend_elements = [
    mpatches.Patch(color='red', label='A (1.0)'),
    mpatches.Patch(color='orange', label='E (0.8)'),
    mpatches.Patch(color='green', label='I (0.5)'),
    mpatches.Patch(color='blue', label='O (0.3)'),
]
ax2.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# =============================================
# STATISTIK
# =============================================
print("\n" + "="*60)
print("FINAL STATISTICS")
print("="*60)

used_area = sum(p['p'] * p['l'] for p in best_placements)
building_area = bldg_l * bldg_w
utilisation = used_area / building_area * 100

print(f"Building: {bldg_l:.1f}m × {bldg_w:.1f}m = {building_area:.1f} m²")
print(f"Used area: {used_area:.1f} m²")
print(f"Utilisation: {utilisation:.1f}%")
print(f"Circulation: {building_area - used_area:.1f} m²")

rotated = [p for p in best_placements if p['rotated']]
print(f"\nRotated rooms: {len(rotated)}")
for p in rotated:
    print(f"   - {p['name']}")

print("\n🔗 Top 5 Strongest Relationships (A/E):")
strong_flows = [(i, j, W[i,j]) for i in range(n) for j in range(i+1, n) if W[i,j] >= 0.8]
strong_flows.sort(key=lambda x: x[2], reverse=True)
for i, j, w in strong_flows[:5]:
    x1, y1 = pos_center[i]
    x2, y2 = pos_center[j]
    dist = abs(x1 - x2) + abs(y1 - y2)
    print(f"   {room_names[i]} ↔ {room_names[j]}: weight={w:.1f}, distance={dist:.1f}m")
```

```python
import numpy as np
import random
import math
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# =============================================
# DATA RUANGAN (dengan allowance 50%)
# =============================================
rooms_spec = [
    (1, "Direktur", 43.5, 1.33),
    (2, "Keuangan", 43.5, 1.67),
    (3, "SDM", 43.5, 1.25),
    (4, "Produksi", 40.5, 1.67),
    (5, "QC", 36.0, 1.67),
    (6, "Pemasaran", 36.0, 1.67),
    (7, "R&D", 28.5, 2.50),
    (8, "Rapat", 72.0, 1.33),
    (9, "Arsip", 22.5, 1.67),
    (10, "Resepsionis", 24.0, 1.00),
    (11, "Musholla", 30.0, 1.25),
    (12, "Pantry", 24.0, 1.00),
    (13, "Keamanan", 13.5, 1.00),
    (14, "Kesehatan", 13.5, 1.00),
    (15, "Toilet W", 4.5, 0.75),
    (16, "Toilet P", 4.5, 0.75),
    (17, "Sink", 3.6, 0.833)
]

rooms = []
for id_, name, area, aspect in rooms_spec:
    l_dim = np.sqrt(area / aspect)
    p_dim = aspect * l_dim
    rooms.append({
        "id": id_ - 1,
        "name": name,
        "p": p_dim,
        "l": l_dim,
        "area": area
    })

n = len(rooms)
room_names = [r['name'] for r in rooms]
total_area = sum(r['area'] for r in rooms)

# =============================================
# FLOW MATRIX (W)
# =============================================
W = np.zeros((n, n))
arc_edges = [
    (1,2,0.8), (1,3,0.8), (1,4,0.5), (1,5,0.5), (1,6,0.8), (1,7,0.5), (1,8,1.0), (1,9,0.3), (1,10,1.0), (1,13,0.3),
    (2,3,0.5), (2,4,0.5), (2,5,0.3), (2,6,0.5), (2,7,0.3), (2,8,0.5), (2,9,0.8), (2,10,0.3),
    (3,4,0.5), (3,5,0.3), (3,6,0.3), (3,7,0.3), (3,8,0.5), (3,9,0.8), (3,10,0.3), (3,13,0.5), (3,14,0.5), (3,15,0.5),
    (4,5,1.0), (4,6,0.5), (4,7,0.5), (4,8,0.5), (4,9,0.5), (4,14,0.3),
    (5,6,0.3), (5,7,0.5), (5,8,0.5), (5,9,0.5), (5,14,0.3), (5,15,0.3),
    (6,7,0.5), (6,8,0.8), (6,9,0.3), (6,10,0.8), (6,15,0.3),
    (7,8,0.5), (7,9,0.8), (7,10,0.3), (7,15,0.3),
    (8,9,0.5), (8,10,0.5), (8,12,0.8), (8,15,0.5), (8,16,0.5), (8,17,0.3),
    (9,12,-0.02), (9,13,0.3), (9,15,-0.02), (9,16,-0.02), (9,17,-0.02),
    (10,12,0.3), (10,13,1.0), (10,15,0.5), (10,16,0.5), (10,17,0.3),
    (11,15,0.3), (11,16,0.3), (11,17,1.0),
    (12,14,0.3), (12,15,0.5), (12,16,0.5), (12,17,1.0),
    (13,14,0.3),
    (14,15,0.5), (14,16,0.5), (14,17,1.0),
    (15,16,0.5), (15,17,0.5),
    (16,17,0.5)
]
for u, v, w in arc_edges:
    W[u-1, v-1] = w
    W[v-1, u-1] = w

print(f"✅ Data loaded: {n} rooms, total area = {total_area:.1f} m²")

# =============================================
# COST FUNCTION
# =============================================
def calculate_cost(placements):
    pos = {}
    for p in placements:
        idx = p['id']
        pos[idx] = (p['x'] + p['p']/2, p['y'] + p['l']/2)

    total_cost = 0
    for i in range(n):
        for j in range(i+1, n):
            if W[i, j] != 0:
                if i not in pos or j not in pos:
                    continue
                x1, y1 = pos[i]
                x2, y2 = pos[j]
                dist = abs(x1 - x2) + abs(y1 - y2)
                if W[i, j] > 0:
                    total_cost += W[i, j] * dist
                else:
                    total_cost += abs(W[i, j]) * max(0, 20 - dist)
    return total_cost

# =============================================
# INITIAL LAYOUT (Grid-based, pasti muat)
# =============================================
def create_initial_layout(rooms, clearance=0.15):
    """Create a feasible initial layout using grid packing"""
    # Sort by area (largest first)
    rooms_sorted = sorted(rooms, key=lambda r: r['area'], reverse=True)

    # Estimate building size (generous)
    total_area = sum(r['area'] for r in rooms)
    aspect = 1.4
    bldg_w = np.sqrt(total_area * 1.25 / aspect)  # 25% extra space
    bldg_l = bldg_w * aspect

    placements = []
    shelves = []

    for room in rooms_sorted:
        placed = False

        # Try existing shelves
        for shelf in shelves:
            y_start, x_end, shelf_h = shelf
            if room['l'] <= shelf_h and x_end + room['p'] + clearance <= bldg_l:
                placements.append({
                    'id': room['id'],
                    'name': room['name'],
                    'x': x_end,
                    'y': y_start,
                    'p': room['p'],
                    'l': room['l'],
                    'rotated': False
                })
                shelf[1] = x_end + room['p'] + clearance
                placed = True
                break
            elif room['p'] <= shelf_h and x_end + room['l'] + clearance <= bldg_l:
                placements.append({
                    'id': room['id'],
                    'name': room['name'],
                    'x': x_end,
                    'y': y_start,
                    'p': room['l'],
                    'l': room['p'],
                    'rotated': True
                })
                shelf[1] = x_end + room['l'] + clearance
                placed = True
                break

        if placed:
            continue

        # New shelf
        if not shelves:
            new_y = clearance
        else:
            new_y = max(s[0] + s[2] for s in shelves) + clearance

        if new_y + room['l'] + clearance <= bldg_w:
            shelves.append([new_y, clearance, room['l']])
            placements.append({
                'id': room['id'],
                'name': room['name'],
                'x': clearance,
                'y': new_y,
                'p': room['p'],
                'l': room['l'],
                'rotated': False
            })
            shelves[-1][1] = clearance + room['p'] + clearance
        elif new_y + room['p'] + clearance <= bldg_w:
            shelves.append([new_y, clearance, room['p']])
            placements.append({
                'id': room['id'],
                'name': room['name'],
                'x': clearance,
                'y': new_y,
                'p': room['l'],
                'l': room['p'],
                'rotated': True
            })
            shelves[-1][1] = clearance + room['l'] + clearance
        else:
            # Expand building vertically
            bldg_w = new_y + max(room['l'], room['p']) + clearance * 2
            shelves.append([new_y, clearance, room['l']])
            placements.append({
                'id': room['id'],
                'name': room['name'],
                'x': clearance,
                'y': new_y,
                'p': room['p'],
                'l': room['l'],
                'rotated': False
            })
            shelves[-1][1] = clearance + room['p'] + clearance

    # Calculate actual building bounds
    max_x = max(p['x'] + p['p'] for p in placements) + clearance
    max_y = max(p['y'] + p['l'] for p in placements) + clearance

    return placements, max_x, max_y

# =============================================
# COMPRESS LAYOUT (SA with shrinking)
# =============================================
def swap_rooms(placements, i, j):
    new_placements = []
    for p in placements:
        new_p = p.copy()
        if p['id'] == i:
            for q in placements:
                if q['id'] == j:
                    new_p['x'] = q['x']
                    new_p['y'] = q['y']
                    new_p['p'] = q['p']
                    new_p['l'] = q['l']
                    new_p['rotated'] = q.get('rotated', False)
                    break
        elif p['id'] == j:
            for q in placements:
                if q['id'] == i:
                    new_p['x'] = q['x']
                    new_p['y'] = q['y']
                    new_p['p'] = q['p']
                    new_p['l'] = q['l']
                    new_p['rotated'] = q.get('rotated', False)
                    break
        new_placements.append(new_p)
    return new_placements

def compress_layout(initial_placements, initial_l, initial_w, target_area, max_iter=5000):
    current = initial_placements
    current_l, current_w = initial_l, initial_w
    current_cost = calculate_cost(current)

    best = current
    best_cost = current_cost
    best_l, best_w = current_l, current_w

    T = 100.0
    alpha = 0.998

    for iteration in range(max_iter):
        # Either swap rooms or shrink building
        if random.random() < 0.7:
            # Swap
            i, j = random.sample(range(n), 2)
            new = swap_rooms(current, i, j)
            new_l, new_w = current_l, current_w
        else:
            # Shrink
            shrink_factor = random.uniform(0.98, 0.995)
            new_l = current_l * shrink_factor
            new_w = current_w * shrink_factor

            # Don't shrink below target area
            if new_l * new_w < target_area:
                new_l = np.sqrt(target_area * (current_l / current_w))
                new_w = target_area / new_l

            # Try to fit
            new = []
            for p in current:
                new_p = p.copy()
                # Scale positions
                new_p['x'] = p['x'] * (new_l / current_l)
                new_p['y'] = p['y'] * (new_w / current_w)
                new_p['p'] = p['p']  # Keep same size
                new_p['l'] = p['l']
                new.append(new_p)

        # Check if valid (no overlap, within bounds)
        valid = True
        for i, p1 in enumerate(new):
            if p1['x'] < 0 or p1['y'] < 0:
                valid = False
                break
            if p1['x'] + p1['p'] > new_l or p1['y'] + p1['l'] > new_w:
                valid = False
                break
            for j, p2 in enumerate(new):
                if i >= j:
                    continue
                if not (p1['x'] + p1['p'] <= p2['x'] or p2['x'] + p2['p'] <= p1['x'] or
                        p1['y'] + p1['l'] <= p2['y'] or p2['y'] + p2['l'] <= p1['y']):
                    valid = False
                    break
            if not valid:
                break

        if not valid:
            continue

        new_cost = calculate_cost(new)
        delta = new_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = new
            current_cost = new_cost
            current_l, current_w = new_l, new_w

            if current_cost < best_cost:
                best = current
                best_cost = current_cost
                best_l, best_w = current_l, current_w

        T *= alpha

        if iteration % 1000 == 0:
            area = current_l * current_w
            print(f"   Iter {iteration}: Cost={current_cost:.1f}, Area={area:.1f}m², Util={total_area/area*100:.1f}%")

    return best, best_cost, best_l, best_w

# =============================================
# RUN
# =============================================
print("\n" + "="*60)
print("🎯 LAYOUT COMPRESSION (Target: 483.6 m²)")
print("="*60)

# Create initial feasible layout
placements, bldg_l, bldg_w = create_initial_layout(rooms, clearance=0.15)
initial_cost = calculate_cost(placements)

print(f"\n✅ Initial feasible layout:")
print(f"   Building: {bldg_l:.2f}m × {bldg_w:.2f}m = {bldg_l * bldg_w:.1f} m²")
print(f"   Utilisation: {total_area / (bldg_l * bldg_w) * 100:.1f}%")
print(f"   Cost: {initial_cost:.2f}")

print("\n🔥 Compressing layout...")
best_placements, best_cost, final_l, final_w = compress_layout(
    placements, bldg_l, bldg_w, total_area, max_iter=5000
)

print(f"\n✅ Final compressed layout:")
print(f"   Building: {final_l:.2f}m × {final_w:.2f}m = {final_l * final_w:.1f} m²")
print(f"   Utilisation: {total_area / (final_l * final_w) * 100:.1f}%")
print(f"   Cost: {best_cost:.2f}")

# =============================================
# VISUALIZATION
# =============================================
fig, ax = plt.subplots(1, 1, figsize=(14, 10))
colors = plt.cm.tab20(np.linspace(0, 1, n))

for p in best_placements:
    i = p['id']
    rect = patches.Rectangle(
        (p['x'], p['y']),
        p['p'], p['l'],
        linewidth=2, edgecolor='black',
        facecolor=colors[i], alpha=0.8
    )
    ax.add_patch(rect)

    short_name = p['name'].replace('Departemen ', '')[:10]
    label = f"{i+1}\n{short_name}"
    if p.get('rotated', False):
        label += "\n↻"
    ax.text(p['x'] + p['p']/2, p['y'] + p['l']/2, label,
             ha='center', va='center', fontsize=8, fontweight='bold')

# Flow lines
pos_center = {p['id']: (p['x'] + p['p']/2, p['y'] + p['l']/2) for p in best_placements}

edges_flow = []
for i in range(n):
    for j in range(i+1, n):
        if abs(W[i, j]) > 0.3:
            edges_flow.append((i, j, W[i, j]))

edges_flow.sort(key=lambda x: abs(x[2]), reverse=True)

for i, j, w in edges_flow[:10]:
    if w >= 1.0:
        color, width, alpha = 'red', 3.0, 0.6
    elif w >= 0.8:
        color, width, alpha = 'orange', 2.5, 0.5
    elif w >= 0.5:
        color, width, alpha = 'green', 2.0, 0.4
    elif w > 0:
        color, width, alpha = 'blue', 1.5, 0.3
    else:
        color, width, alpha = 'black', 1.5, 0.4

    if i in pos_center and j in pos_center:
        x1, y1 = pos_center[i]
        x2, y2 = pos_center[j]
        ax.plot([x1, x2], [y1, y2],
                 color=color, linewidth=width, alpha=alpha, zorder=1)

ax.set_xlim(-0.5, final_l + 0.5)
ax.set_ylim(-0.5, final_w + 0.5)
ax.set_aspect('equal')
ax.set_title(f'Compressed Layout\nBuilding: {final_l:.1f}×{final_w:.1f}m | Area: {final_l*final_w:.1f}m² | Cost: {best_cost:.1f}',
             fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)

import matplotlib.patches as mpatches
legend_elements = [
    mpatches.Patch(color='red', label='A (1.0)'),
    mpatches.Patch(color='orange', label='E (0.8)'),
    mpatches.Patch(color='green', label='I (0.5)'),
    mpatches.Patch(color='blue', label='O (0.3)'),
    mpatches.Patch(color='black', label='X (-0.02)'),
]
ax.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()

# =============================================
# STATISTIK
# =============================================
print("\n" + "="*60)
print("📊 FINAL STATISTICS")
print("="*60)

used_area = sum(p['p'] * p['l'] for p in best_placements)
building_area = final_l * final_w
utilisation = used_area / building_area * 100

print(f"Building: {final_l:.2f}m × {final_w:.2f}m = {building_area:.2f} m²")
print(f"Used area: {used_area:.2f} m²")
print(f"Target area: {total_area:.2f} m²")
print(f"Utilisation: {utilisation:.1f}%")

print("\n🔗 Top 5 Strongest Relationships (A/E):")
strong_flows = [(i, j, W[i,j]) for i in range(n) for j in range(i+1, n) if W[i,j] >= 0.8]
strong_flows.sort(key=lambda x: x[2], reverse=True)
for i, j, w in strong_flows[:5]:
    if i in pos_center and j in pos_center:
        x1, y1 = pos_center[i]
        x2, y2 = pos_center[j]
        dist = abs(x1 - x2) + abs(y1 - y2)
        print(f"   {room_names[i]} ↔ {room_names[j]}: weight={w:.1f}, distance={dist:.1f}m")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import random
import math

# Data ruangan dari tabel
rooms_data = [
    {"name": "Direktur", "sub": ["Direktur", "Sekretaris"], "areas": [30, 13.5]},
    {"name": "Keuangan & Akuntansi", "sub": ["Kepala Dept. Keu", "Keuangan & Anggaran", "Purchasing"], "areas": [13.5, 22.5, 7.5]},
    {"name": "SDM & Umum", "sub": ["Kepala Dept. HR", "Staff HR"], "areas": [13.5, 30]},
    {"name": "Produksi & PPIC", "sub": ["Kepala Dept. Produksi", "Staff Produksi & PPIC"], "areas": [18, 22.5]},
    {"name": "Quality Control", "sub": ["Kepala Dept. QC", "Staff QC"], "areas": [13.5, 22.5]},
    {"name": "Pemasaran & Penjualan", "sub": ["Kepala Dept. Marketing", "Staff Marketing & Sales"], "areas": [13.5, 22.5]},
    {"name": "R&D dan Desain Produk", "sub": ["Kepala Dept. R&D", "Staff R&D"], "areas": [13.5, 15]},
    {"name": "Ruang Rapat Besar", "sub": ["Ruang Rapat"], "areas": [72]},
    {"name": "Arsip & Dokumentasi", "sub": ["Arsip"], "areas": [22.5]},
    {"name": "Tamu & Resepsionis", "sub": ["Resepsionis"], "areas": [24]},
    {"name": "Musholla", "sub": ["Musholla"], "areas": [30]},
    {"name": "Istirahat & Pantry", "sub": ["Pantry"], "areas": [24]},
    {"name": "Pos Keamanan", "sub": ["Satpam"], "areas": [13.5]},
    {"name": "Ruang Kesehatan", "sub": ["Kesehatan"], "areas": [13.5]},
    {"name": "Toilet Wanita", "sub": ["Toilet W1", "Toilet W2"], "areas": [2.25, 2.25]},
    {"name": "Toilet Pria", "sub": ["Toilet P1", "Toilet P2"], "areas": [2.25, 2.25]},
    {"name": "Sink", "sub": ["Sink 1", "Sink 2"], "areas": [1.8, 1.8]},
]

# Total luas maksimal (target)
TOTAL_AREA = 483.6
BUILDING_WIDTH = math.sqrt(TOTAL_AREA) * 1.2  # Bentuk persegi panjang
BUILDING_HEIGHT = TOTAL_AREA / BUILDING_WIDTH

# Flatten semua sub-ruangan
all_subrooms = []
for room in rooms_data:
    for sub_name, area in zip(room["sub"], room["areas"]):
        all_subrooms.append({
            "department": room["name"],
            "name": sub_name,
            "area": area
        })

# Warna untuk tiap departemen
color_map = plt.cm.Set3(np.linspace(0, 1, len(rooms_data)))
dept_colors = {room["name"]: color_map[i] for i, room in enumerate(rooms_data)}

def calculate_dimensions(area, max_aspect_ratio=3.0):
    """Hitung dimensi ruangan dengan aspect ratio yang wajar"""
    min_dim = math.sqrt(area / max_aspect_ratio)
    max_dim = math.sqrt(area * max_aspect_ratio)

    if random.random() < 0.5:
        width = random.uniform(min_dim, max_dim)
        height = area / width
    else:
        height = random.uniform(min_dim, max_dim)
        width = area / height

    return width, height

def generate_initial_layout():
    """Generate layout awal dengan packing sederhana"""
    layout = []
    x, y = 0.0, 0.0
    current_row_height = 0.0
    row_items = []

    # Sort by area (largest first) untuk packing lebih baik
    sorted_rooms = sorted(all_subrooms, key=lambda r: r["area"], reverse=True)

    for room in sorted_rooms:
        width, height = calculate_dimensions(room["area"])

        # Check if fits in current row
        if x + width > BUILDING_WIDTH:
            # Move to next row
            y += current_row_height
            x = 0.0
            current_row_height = 0.0

            # If still doesn't fit, adjust dimensions
            if y >= BUILDING_HEIGHT:
                # Reset to fit in building
                y = 0.0
                width = min(width, BUILDING_WIDTH)
                height = room["area"] / width

        # Check if exceeds building height
        if y + height > BUILDING_HEIGHT:
            # Adjust height to fit, ensure it's not zero
            if y < BUILDING_HEIGHT:
                height = max(0.1, BUILDING_HEIGHT - y)  # Ensure height > 0
                width = room["area"] / height
            else:
                # If y is already beyond building, start new column
                x += current_row_height  # Use height as width for new column
                y = 0.0
                height = min(math.sqrt(room["area"]), BUILDING_HEIGHT)
                width = room["area"] / height

        # Ensure dimensions are valid
        if height <= 0 or width <= 0:
            width = math.sqrt(room["area"])
            height = width

        layout.append({
            **room,
            "x": x,
            "y": y,
            "width": width,
            "height": height
        })

        x += width
        current_row_height = max(current_row_height, height)

    # Fix any rooms that might be out of bounds
    for room in layout:
        room["x"] = max(0, min(room["x"], BUILDING_WIDTH - room["width"]))
        room["y"] = max(0, min(room["y"], BUILDING_HEIGHT - room["height"]))

    return layout

def check_overlap(room1, room2):
    """Check if two rooms overlap"""
    return not (room1["x"] + room1["width"] <= room2["x"] or
                room2["x"] + room2["width"] <= room1["x"] or
                room1["y"] + room1["height"] <= room2["y"] or
                room2["y"] + room2["height"] <= room1["y"])

def calculate_overlap_area(room1, room2):
    """Calculate overlap area between two rooms"""
    x_overlap = max(0, min(room1["x"] + room1["width"], room2["x"] + room2["width"]) -
                       max(room1["x"], room2["x"]))
    y_overlap = max(0, min(room1["y"] + room1["height"], room2["y"] + room2["height"]) -
                       max(room1["y"], room2["y"]))
    return x_overlap * y_overlap

def calculate_cost(layout):
    """Hitung cost function untuk layout"""
    cost = 0

    # 1. Penalty untuk overlap
    for i, room1 in enumerate(layout):
        for j, room2 in enumerate(layout[i+1:], i+1):
            if check_overlap(room1, room2):
                overlap_area = calculate_overlap_area(room1, room2)
                cost += overlap_area * 10000  # Very heavy penalty for overlap

    # 2. Penalty untuk keluar dari boundary
    for room in layout:
        if room["x"] < 0:
            cost += abs(room["x"]) * 1000
        if room["y"] < 0:
            cost += abs(room["y"]) * 1000
        if room["x"] + room["width"] > BUILDING_WIDTH:
            cost += (room["x"] + room["width"] - BUILDING_WIDTH) * 1000
        if room["y"] + room["height"] > BUILDING_HEIGHT:
            cost += (room["y"] + room["height"] - BUILDING_HEIGHT) * 1000

    # 3. Reward untuk compactness (minimize empty space)
    used_area = sum(r["width"] * r["height"] for r in layout)
    empty_space = max(0, BUILDING_WIDTH * BUILDING_HEIGHT - used_area)
    cost += empty_space * 10

    # 4. Group departments together
    dept_positions = {}
    for room in layout:
        dept = room["department"]
        center = (room["x"] + room["width"]/2, room["y"] + room["height"]/2)
        if dept not in dept_positions:
            dept_positions[dept] = []
        dept_positions[dept].append(center)

    for dept, positions in dept_positions.items():
        if len(positions) > 1:
            # Calculate average distance between rooms in same department
            for i in range(len(positions)):
                for j in range(i+1, len(positions)):
                    dist = math.sqrt((positions[i][0] - positions[j][0])**2 +
                                   (positions[i][1] - positions[j][1])**2)
                    cost += dist * 20  # Penalty for separation

    # 5. Penalty untuk aspect ratio yang terlalu ekstrem
    for room in layout:
        aspect_ratio = max(room["width"], room["height"]) / min(room["width"], room["height"])
        if aspect_ratio > 4:
            cost += (aspect_ratio - 4) * 100

    return cost

def simulated_annealing(initial_temp=5000, cooling_rate=0.995, min_temp=0.1, iterations_per_temp=50):
    """Optimasi layout menggunakan Simulated Annealing"""

    current_layout = generate_initial_layout()
    current_cost = calculate_cost(current_layout)

    best_layout = [room.copy() for room in current_layout]
    best_cost = current_cost

    temperature = initial_temp
    iteration = 0

    costs_history = []

    print(f"Initial cost: {current_cost:.2f}")

    while temperature > min_temp:
        for _ in range(iterations_per_temp):
            # Generate neighbor solution
            new_layout = [room.copy() for room in current_layout]

            # Randomly modify 1-3 rooms
            num_modifications = random.randint(1, 3)
            modified_indices = random.sample(range(len(new_layout)), min(num_modifications, len(new_layout)))

            for idx in modified_indices:
                room = new_layout[idx]

                # Randomly choose modification type
                modification = random.choice(["move", "resize", "swap"])

                if modification == "move":
                    # Move room
                    room["x"] += random.uniform(-1, 1)
                    room["y"] += random.uniform(-1, 1)

                elif modification == "resize":
                    # Resize room (maintain area)
                    scale = random.uniform(0.9, 1.1)
                    room["width"] *= scale
                    room["height"] = room["area"] / room["width"]

                elif modification == "swap" and len(new_layout) > 1:
                    # Swap positions with another room
                    other_idx = random.choice([i for i in range(len(new_layout)) if i != idx])
                    room2 = new_layout[other_idx]

                    # Swap positions but maintain dimensions
                    room["x"], room2["x"] = room2["x"], room["x"]
                    room["y"], room2["y"] = room2["y"], room["y"]

            # Ensure rooms stay within bounds
            for room in new_layout:
                room["x"] = max(0, min(room["x"], BUILDING_WIDTH - room["width"]))
                room["y"] = max(0, min(room["y"], BUILDING_HEIGHT - room["height"]))

            # Calculate new cost
            new_cost = calculate_cost(new_layout)

            # Accept or reject
            delta_cost = new_cost - current_cost

            if delta_cost < 0 or random.random() < math.exp(-delta_cost / max(temperature, 1e-10)):
                current_layout = new_layout
                current_cost = new_cost

                if current_cost < best_cost:
                    best_layout = [room.copy() for room in current_layout]
                    best_cost = current_cost

        costs_history.append(current_cost)
        temperature *= cooling_rate
        iteration += 1

        if iteration % 50 == 0:
            print(f"Iteration {iteration}, Temperature: {temperature:.2f}, Cost: {current_cost:.2f}, Best: {best_cost:.2f}")

    return best_layout, best_cost, costs_history

def visualize_layout(layout, title="Optimized Floor Plan"):
    """Visualisasi layout ruangan"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

    # Main layout visualization
    ax1.set_xlim(-1, BUILDING_WIDTH + 1)
    ax1.set_ylim(-1, BUILDING_HEIGHT + 1)
    ax1.set_aspect('equal')
    ax1.set_title(title, fontsize=14, fontweight='bold')
    ax1.set_xlabel("Width (m)")
    ax1.set_ylabel("Height (m)")

    # Draw building boundary
    building_rect = patches.Rectangle((0, 0), BUILDING_WIDTH, BUILDING_HEIGHT,
                                      linewidth=2, edgecolor='black', facecolor='none', linestyle='--')
    ax1.add_patch(building_rect)

    # Draw each room
    for room in layout:
        color = dept_colors[room["department"]]

        # Main room rectangle
        rect = patches.Rectangle((room["x"], room["y"]), room["width"], room["height"],
                                linewidth=1.5, edgecolor='black', facecolor=color, alpha=0.7)
        ax1.add_patch(rect)

        # Add room label
        center_x = room["x"] + room["width"] / 2
        center_y = room["y"] + room["height"] / 2

        # Truncate long names
        display_name = room["name"][:20] + ".." if len(room["name"]) > 20 else room["name"]

        ax1.text(center_x, center_y, display_name,
                ha='center', va='center', fontsize=7, fontweight='bold',
                bbox=dict(boxstyle='round,pad=0.2', facecolor='white', alpha=0.8))

        # Add area text
        ax1.text(center_x, center_y - room["height"]/4, f"{room['area']:.1f}m²",
                ha='center', va='center', fontsize=6, style='italic')

    # Legend for departments
    legend_elements = []
    for dept, color in dept_colors.items():
        if any(r["department"] == dept for r in layout):
            legend_elements.append(patches.Patch(facecolor=color, edgecolor='black',
                                                label=dept, alpha=0.7))

    ax1.legend(handles=legend_elements, loc='upper left', bbox_to_anchor=(1.02, 1),
              title="Departments", fontsize=7, title_fontsize=8)

    # Statistics subplot
    ax2.axis('off')

    dept_areas = {}
    for room in layout:
        dept = room["department"]
        area = room["width"] * room["height"]
        if dept not in dept_areas:
            dept_areas[dept] = 0
        dept_areas[dept] += area

    stats_text = f"""
    OPTIMIZATION STATISTICS

    Building Dimensions: {BUILDING_WIDTH:.1f}m × {BUILDING_HEIGHT:.1f}m
    Total Building Area: {BUILDING_WIDTH * BUILDING_HEIGHT:.1f} m²
    Target Total Area: {TOTAL_AREA:.1f} m²

    Number of Rooms: {len(layout)}
    Number of Departments: {len(set(r['department'] for r in layout))}

    Final Cost: {best_cost:.2f}

    Area Distribution by Department:
    """

    for dept, area in sorted(dept_areas.items(), key=lambda x: x[1], reverse=True):
        stats_text += f"\n  • {dept[:30]}: {area:.1f} m²"

    ax2.text(0.1, 0.9, stats_text, transform=ax2.transAxes, fontsize=8,
            verticalalignment='top', fontfamily='monospace',
            bbox=dict(boxstyle='round', facecolor='lightgray', alpha=0.5))

    plt.tight_layout()
    return fig

# Run optimization
print("="*80)
print("SIMULATED ANNEALING FLOOR PLAN OPTIMIZATION")
print("="*80)
print(f"Building Dimensions: {BUILDING_WIDTH:.1f}m x {BUILDING_HEIGHT:.1f}m")
print(f"Total Target Area: {TOTAL_AREA:.1f} m²")
print(f"Number of Rooms: {len(all_subrooms)}")
print("="*80 + "\n")

best_layout, best_cost, costs_history = simulated_annealing(
    initial_temp=10000,
    cooling_rate=0.998,
    min_temp=0.01,
    iterations_per_temp=40
)

print(f"\n✅ Optimization Complete!")
print(f"📊 Final Cost: {best_cost:.2f}")

# Visualize result
fig = visualize_layout(best_layout, "Optimized Office Layout - Simulated Annealing")
plt.show()

# Plot cost history
fig2, ax = plt.subplots(figsize=(10, 6))
ax.plot(costs_history, linewidth=2, color='blue')
ax.set_xlabel("Iteration", fontsize=12)
ax.set_ylabel("Cost", fontsize=12)
ax.set_title("Simulated Annealing - Cost Convergence", fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.set_yscale('log')  # Log scale for better visualization
plt.tight_layout()
plt.show()

# Print detailed room information
print("\n" + "="*80)
print("DETAILED ROOM LAYOUT")
print("="*80)
print(f"{'Department':<35} {'Room':<25} {'Area (m²)':<10} {'Dimensions':<20}")
print("-"*80)
for room in sorted(best_layout, key=lambda x: (x["department"], -x["area"])):
    print(f"{room['department']:<35} {room['name']:<25} {room['area']:<10.1f} "
          f"{room['width']:.1f}m × {room['height']:.1f}m")

# Calculate and show statistics
total_actual_area = sum(r["width"] * r["height"] for r in best_layout)
print("\n" + "="*80)
print("FINAL STATISTICS")
print("="*80)
print(f"Total Target Area: {TOTAL_AREA:.1f} m²")
print(f"Total Actual Area: {total_actual_area:.1f} m²")
print(f"Area Utilization: {(total_actual_area / (BUILDING_WIDTH * BUILDING_HEIGHT)) * 100:.1f}%")
print(f"Empty Space: {BUILDING_WIDTH * BUILDING_HEIGHT - total_actual_area:.1f} m²")
```
